````python
# Learn python OOPS concepts
# https://www.geeksforgeeks.org/python-oops-concepts/

class Parent:
    def get(self):
        print("Parent get")

    def set(self, value_a, value_b):
        self.value_a = value_a
        self.value_b = value_b
        print(f"Parent set value_a: {self.value_a} value_b: {self.value_b}")

class Child(Parent):
    def get(self):
        print(f"Child get: {self.value_a} {self.value_b}")
        
obj = Child()
obj.set(10,20)
obj.get()
````

````python
# Python Inheritance
# Inheritance allows a class (child class) to acquire properties and methods of another class (parent class). It supports hierarchical classification and promotes code reuse.

# Types of Inheritance:
# Single Inheritance: A child class inherits from a single parent class.
# Multiple Inheritance: A child class inherits from more than one parent class.
# Multilevel Inheritance: A child class inherits from a parent class, which in turn inherits from another class.
# Hierarchical Inheritance: Multiple child classes inherit from a single parent class.
# Hybrid Inheritance: A combination of two or more types of inheritance.

class Dog:
    """
    parent class
    """
    def __init__(self, name):
        self.name = name
        print(f"{Dog.__name__} class initialized")

    def display_name(self):
        print(f"Dog name: {self.name}")

class Labrador(Dog):
    """
    child class: single inheritance
    """
    # def __init__(self, name):
    #     super().__init__(name)
    #     print(f"{Labrador.__name__} class initialized")

    def sound(self):
        print(f"{self.name} says Woof!")


labrador = Labrador("Buddy")
labrador.display_name()
labrador.sound()
````

In Python, when you define a subclass (e.g., `Labrador`) that inherits from a superclass (e.g., `Dog`), the `__init__` method of the superclass is not automatically called unless you explicitly invoke it. However, if the subclass does not define its own `__init__` method, Python will automatically use the `__init__` method of the superclass.

Here’s an example to clarify:



````python
class Dog:
    def __init__(self, name):
        self.name = name

class Labrador(Dog):
    pass  # No __init__ method defined here

# Creating an instance of Labrador
labrador = Labrador("Buddy")
print(labrador.name)  # Output: Buddy
````



### Why does this work?
- Since `Labrador` does not define its own `__init__` method, Python automatically uses the `__init__` method of the `Dog` class (its superclass).
- This behavior ensures that the `Dog` class's `__init__` method is used to initialize the `Labrador` instance.

### What happens if `Labrador` defines its own `__init__`?
If `Labrador` defines its own `__init__` method, the superclass's `__init__` method will **not** be called automatically. You would need to explicitly call it using `super()` if you want to initialize the superclass:



````python
class Labrador(Dog):
    def __init__(self, name, color):
        super().__init__(name)  # Explicitly call the superclass's __init__
        self.color = color

labrador = Labrador("Buddy", "Golden")
print(labrador.name)  # Output: Buddy
print(labrador.color)  # Output: Golden
````



### Summary
- If the subclass does not define an `__init__` method, the superclass's `__init__` is automatically used.
- If the subclass defines its own `__init__`, you must explicitly call the superclass's `__init__` using `super()` if needed.

````python
class Dog:
    """
    parent class
    """
    def __init__(self, name):
        self.name = name
        print(f"{Dog.__name__} class initialized")

    def display_name(self):
        print(f"Dog name: {self.name}")

class Labrador(Dog):
    """
    child class: single inheritance
    """
    def __init__(self, name):
        # super().__init__(name)
        print(f"{Labrador.__name__} class initialized")
        self.name = name

    def sound(self):
        print(f"{self.name} says Woof!")


labrador = Labrador("Buddy")
labrador.display_name()
labrador.sound()
````

### ✅ Hierarchical Inheritance

**Definition:** Multiple child classes inherit from a single parent class.

#### 📌 Example:

```python
class Animal:
    def speak(self):
        print("Animal speaks")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

class Cat(Animal):
    def meow(self):
        print("Cat meows")

# Usage
dog = Dog()
dog.speak()  # Inherited from Animal
dog.bark()

cat = Cat()
cat.speak()  # Inherited from Animal
cat.meow()
```

🧠 **Explanation:**

* `Animal` is the **parent class**.
* `Dog` and `Cat` are **child classes** inheriting from the same parent.

---

### ✅ Hybrid Inheritance

**Definition:** A combination of more than one type of inheritance. For example, combining **multiple** and **hierarchical** inheritance.

#### 📌 Example:

```python
class Person:
    def show(self):
        print("Person details")

class Employee(Person):  # Single inheritance
    def work(self):
        print("Employee working")

class Student(Person):  # Hierarchical inheritance
    def study(self):
        print("Student studying")

class Intern(Employee, Student):  # Multiple inheritance
    def intern_work(self):
        print("Intern working")

# Usage
intern = Intern()
intern.show()       # From Person
intern.work()       # From Employee
intern.study()      # From Student
intern.intern_work()
```

🧠 **Explanation:**

* `Employee` and `Student` both inherit from `Person` → **Hierarchical**.
* `Intern` inherits from both `Employee` and `Student` → **Multiple**.
* The combination is called **Hybrid Inheritance**.

Would you like to see how Python resolves method conflicts in hybrid inheritance (i.e., Method Resolution Order)?


````python
#single inheritance
class Dog:
    def __init__(self, name):
        self.name = name

    def display_name(self):
        print(f"Dog's name: {self.name}")

class Labrador(Dog): # single inheritance
    def sound(self):
        print(f"Labrador woofs!")
    
# Multilevel Inheritance
class GuideDog(Labrador):
    def guide(self):
        print(f"{self.name} guides the way!")

# multiple inheritance
class Friendly:
    def greet(self):
        print("Friendly!")

class GoldenRetriever(Dog, Friendly): # Multiple inheritance
    def sound(self):
        print(f"Golden Retriever barks!")

    def __repr__(self):
        return f"{self.name} is a Golden Retriever."

    def __str__(self):
        return f"{self.name} is a friendly golden retriever."


# Creating instances
labrador = Labrador("Buddy")
labrador.display_name()  # Output: Dog's name: Buddy
labrador.sound()  # Output: Labrador woofs!

guide_dog = GuideDog("Max")
guide_dog.display_name()  # Output: Dog's name: Max
guide_dog.guide()  # Output: Max guides the way!

golden_retriever = GoldenRetriever("Charlie")
golden_retriever.display_name()  # Output: Dog's name: Charlie
golden_retriever.greet()  # Output: Friendly!
golden_retriever.sound()  # Output: Golden Retriever barks!

golden_retriever
# str(golden_retriever)
print(golden_retriever)

````

````python
# Python polymorphism
# Polymorphism allows methods to do different things based on the object it is acting upon, even if they share the same name.
# It can be achieved through method overriding and operator overloading.
# Method Overriding: A child class can provide a specific implementation of a method that is already defined in its parent class.
# Operator Overloading: Python allows operators to be redefined and used in different ways for user-defined classes.

# Types of Polymorphism
# Compile-Time Polymorphism: This type of polymorphism is determined during the compilation of the program. It allows methods or operators with the same name to behave differently based on their input parameters or usage. It is commonly referred to as method or operator overloading.
# Run-Time Polymorphism: This type of polymorphism is determined during the execution of the program. It occurs when a subclass provides a specific implementation for a method already defined in its parent class, commonly known as method overriding.

# parent class
class Dog:
    def sound(self):
        print("Dog sound") # default implementation

# Run time polymorphism: method overriding
class Labrador(Dog):
    def sound(self):
        print("Labrador woofs!") # overridden implementation

class Beagle(Dog):
    def sound(self):
        print("Beagle barks!") # overriding parent method

# Compile time polymorphism: method overloading mimic
class Calculator:
    def add(self, a, b=0, c=0):
        return a + b + c # supports multiple ways to call add()
    
# Run-time polymorphism: method overriding
dogs = [Dog(), Labrador(), Beagle()]
for dog in dogs:
    dog.sound() # Output: Dog sound, Labrador woofs!, Beagle barks!

# Compile time polymorphism(mimiced using default arguments)
calculator = Calculator()
print(calculator.add(5)) # Output: 5 one argument
print(calculator.add(5, 10)) # Output: 15 two arguments
print(calculator.add(5, 10, 15)) # Output: 30 three arguments
````

No, **method overloading** (as seen in languages like Java or C++) is not directly supported in Python. In Python, you cannot define multiple methods with the same name but different parameter lists in the same class. If you try to do so, the last method definition will overwrite the previous ones.

### Example of Overwriting:


````python
class Example:
    def greet(self, name):
        print(f"Hello, {name}!")

    def greet(self):  # This will overwrite the previous method
        print("Hello!")

obj = Example()
obj.greet()  # Output: Hello!
````



### How to Achieve Method Overloading in Python?
Python provides flexibility through **default arguments**, **variable-length arguments (`*args`, `**kwargs`)**, and **type checking** to simulate method overloading.

#### Using Default Arguments:


````python
class Example:
    def greet(self, name=None):
        if name:
            print(f"Hello, {name}!")
        else:
            print("Hello!")

obj = Example()
obj.greet()         # Output: Hello!
obj.greet("Alice")  # Output: Hello, Alice!
````



#### Using `*args` and `**kwargs`:


````python
class Example:
    def greet(self, *args):
        if len(args) == 0:
            print("Hello!")
        elif len(args) == 1:
            print(f"Hello, {args[0]}!")
        else:
            print("Hello, everyone!")

obj = Example()
obj.greet()               # Output: Hello!
obj.greet("Alice")        # Output: Hello, Alice!
obj.greet("Alice", "Bob") # Output: Hello, everyone!
````



#### Using Type Checking:


````python
class Example:
    def greet(self, name):
        if isinstance(name, str):
            print(f"Hello, {name}!")
        elif isinstance(name, list):
            print(f"Hello, {', '.join(name)}!")
        else:
            print("Hello!")

obj = Example()
obj.greet("Alice")          # Output: Hello, Alice!
obj.greet(["Alice", "Bob"]) # Output: Hello, Alice, Bob!
````



### Summary
While Python does not support method overloading directly, you can achieve similar functionality using default arguments, `*args`/`**kwargs`, or type checking. This approach aligns with Python's philosophy of simplicity and flexibility.
